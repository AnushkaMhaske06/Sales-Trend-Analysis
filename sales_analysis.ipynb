```python
import pandas as pd

df = pd.read_excel("sample_-_superstore.xls")

df.head()
```

```python
print(df.shape)
print(df.columns)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

print(df["Order Date"].dtype)
print(df["Order Date"].min())
print(df["Order Date"].max())
```

```python
df["Month"] = df["Order Date"].dt.to_period("M")

monthly_sales = df.groupby("Month")["Sales"].sum()

monthly_sales.head(10)
```

```python
import matplotlib.pyplot as plt

monthly_sales.plot(kind="line", figsize=(12, 5), marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
import os

os.makedirs("charts", exist_ok=True)

monthly_sales.plot(kind="line", figsize=(12, 5), marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.savefig("charts/monthly_sales_trend.png", dpi=300)
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
category_sales.plot(
    kind="bar",
    figsize=(8, 5),
    title="Sales by Category"
)

plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
os.makedirs("charts", exist_ok=True)

category_sales.plot(
    kind="bar",
    figsize=(8, 5),
    title="Sales by Category"
)

plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("charts/category_sales.png", dpi=300)
plt.show()
```

```python
category_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7),
    title="Sales Share by Category"
)

plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
category_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7),
    title="Sales Share by Category"
)

plt.ylabel("")
plt.tight_layout()

plt.savefig("charts/category_sales_pie.png", dpi=300)
plt.show()
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
best_category = category_sales.idxmax()

print("SALES ANALYSIS SUMMARY")
print("----------------------")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Best-Selling Category: {best_category}")
print(f"Total Orders: {df['Order ID'].nunique()}")
```

```python
summary = f"""
SALES ANALYSIS SUMMARY
----------------------
Total Sales: ${total_sales:,.2f}
Total Profit: ${total_profit:,.2f}
Best-Selling Category: {best_category}
Total Orders: {df['Order ID'].nunique()}
"""

with open("summary.txt", "w", encoding="utf-8") as file:
    file.write(summary)

print("Summary saved successfully!")
```
